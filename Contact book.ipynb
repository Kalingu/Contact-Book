```python
# A simple contact management system.
# Features: Add, Search, Delete, List, and Exit.
```

```python
contact={}
while True:
    choice=(input("1)add 2)search 3)delete 4)list 5)exit:"))
    if choice == "1":
        name=input("enter your name:")
        number=input("enter your number:")
        contact[name]=number
    elif choice == "2" :
        name=input("enter your name:")
        if name in contact:
            print(contact[name])
        else:
            print("not available")
    elif choice == "3":
        name = input("enter your name:")
        if name in contact:
            del contact[name]
        else:
           print("not available")
    elif choice == "4":
        for name, number in contact.items():
            print(name, number)
    elif choice == "5":
        break
    else:
        print("invalid value")
```
